# Patstāvīgie uzdevumi · NumPy

## 1. uzdevums · Paraugu blīvums

Laboratorijā nosvērti četri materiāla paraugi un noteikts to tilpums. Noskaidro, kuriem paraugiem blīvums pārsniedz **2,72 g/cm³**. Dati ir simulēti.

| Paraugs | Masa (g) | Tilpums (cm³) |
|---|---:|---:|
| A | 26,8 | 10,0 |
| B | 27,4 | 10,0 |
| C | 26,4 | 9,6 |
| D | 27,1 | 10,0 |

```python
import numpy as np
```

### 1. Izveido masīvus

Ar `np.array()` izveido divus masīvus: `masa_g` un `tilpums_cm3`. Abos saglabā secību A, B, C, D, lai masa un tilpums vienā pozīcijā attiektos uz vienu paraugu. Python kodā decimālzīme ir punkts.

```python
masa_g = np.array([26.8, 27.4, 26.4, 27.1])
tilpums_cm3 = np.array([10.0, 10.0, 9.6, 10.0])
```

### 2. Aprēķini katra parauga blīvumu

Blīvums ir masa, dalīta ar tilpumu. Izdali masīvus un saglabā rezultātu masīvā `blivums`. Parādi to. Pašpārbaudei: paraugam A jāiegūst 2,68 g/cm³.

```python
blivums = masa_g / tilpums_cm3
print(blivums)
```

### 3. Raksturo paraugus

Aprēķini četru blīvumu vidējo vērtību un atrodi mazāko un lielāko blīvumu. Izmanto tās pašas trīs funkcijas, kuras mācību piemērā lietoji biezumam.

```python
print("Medium:", np.mean(blivums))
print("Min:", np.min(blivums))
print("Max:", np.max(blivums))
```

### 4. Atlasi paraugus

Izveido masku ar nosacījumu `blivums > 2.72`. Izmanto masku, lai parādītu tikai atlasītās blīvuma vērtības. Pēc maskas un sākotnējās secības nosaki paraugu burtus.

```python
maska = blivums > 2.72
print(maska)
print(blivums[maska])

```

## 2. uzdevums · Temperatūra krāsnī

Trīs karsēšanas ciklos vienlaikus reģistrēja temperatūru divās krāsns vietās. Dati ir simulēti. Vai vidēji karstāks ir pie durvīm vai krāsns dziļumā?

| Cikls | Pie durvīm (°C) | Dziļumā (°C) |
|---|---:|---:|
| A | 790 | 810 |
| B | 798 | 806 |
| C | 794 | 808 |

**1. Izveido 2D masīvu.** Ar `np.array()` ievadi temperatūras skaitļus. Saglabā tabulā doto secību: viena rinda ir viens cikls, viena kolonna ir viena mērījuma vieta. Parādi masīva izmēru. Ar rindas un kolonnas indeksiem atlasi temperatūru ciklā B pie durvīm. Pašpārbaudei jāiegūst 798 °C.

```python
temperaturas = np.array([
    [790, 810],
    [798, 806],
    [794, 808],
])

print(temperaturas.shape)
print(temperaturas[1, 0])
```

**2. Aprēķini katra cikla vidējo temperatūru.** Izmanto `np.mean()` un izvēlies atbilstošo asi. Cik vērtību sagaidi? Pirmajam ciklam jāiegūst 800 °C.

```python
ciklu_videjas = np.mean(temperaturas, axis=1)
print(ciklu_videjas)
```

**3. Salīdzini abas krāsns vietas.** Aprēķini vidējo temperatūru katrā vietā.

```python
pie_durvim, dziluma = np.mean(temperaturas, axis=0)

print("Pie durvīm:", pie_durvim, "°C")
print("Krāsns dziļumā:", dziluma, "°C")
print("Starpība:", dziluma - pie_durvim, "°C")
```
