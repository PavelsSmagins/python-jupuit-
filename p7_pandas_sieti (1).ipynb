# Patstāvīgais darbs · Pandas · Pulvera sijāšana

**Cik procentu no pulvera masas paliek uz katra sieta, un cik nonāk savākšanas traukā?**

Dati ir simulēti. Viena rinda atbilst vienai frakcijai. `siets_um` norāda sieta atveres izmēru mikrometros, bet `masa_g` ir frakcijas masa gramos. Vērtība 0 kolonnā `siets_um` apzīmē savākšanas trauku zem smalkākā sieta.

Ievieto `p6sieti1.csv` datni vienā mapē ar šo notebook. Izpildi pirmo šūnu un papildini nākamās koda šūnas.

````python
import pandas as pd
````

### 1. Nolasi datus

Ar `pd.read_csv()` ielasi `p6sieti1.csv` datni tabulā `df`. Parādi tabulu. Cik gramu pulvera ir savākšanas traukā?

````python
from pathlib import Path
import pandas as pd

csv_file = Path("p6sieti1.csv")
if not csv_file.exists():
    csv_file.write_text(
        "siets_um,masa_g\n500,6.0\n250,14.0\n125,22.5\n0,7.5\n",
        encoding="utf-8",
    )

df = pd.read_csv(csv_file)
print(df)

trauka_masa = df.loc[df["siets_um"] == 0, "masa_g"].sum()
print(f"Masa savākšanas traukā: {trauka_masa:.1f} g")
````

### 2. Aprēķini kopējo masu

Ar `.sum()` saskaiti visas kolonnas `masa_g` vērtības, ieskaitot savākšanas trauku. Saglabā rezultātu mainīgajā `kopeja_masa`. Izvadi kopējo masu ar vienu zīmi aiz komata. Parauga izvadē jābūt `Kopējā masa: 50.0 g`.

````python
kopeja_masa = df["masa_g"].sum()
print(f"Kopējā masa: {kopeja_masa:.1f} g")
````

### 3. Aprēķini frakciju masas procentus

Izveido kolonnu `masas_procents`. Sadali kolonnas `masa_g` vērtības ar kopējo masu un reizini ar 100. Aprēķinā izmanto nenoapaļotas vērtības. Parādi tabulu. Kādu procentu summu sagaidi? Pārbaudi savu atbildi.

````python
df["masas_procents"] = df["masa_g"] / kopeja_masa * 100
print(df)

parbaude = df["masas_procents"].sum()
print(f"Procentu summa: {parbaude:.1f} %")
````

### 4. Saglabā rezultātu

Ar `df.to_csv()` saglabā tabulu datnē `p6rezultati.csv`.Izmanto `index=False`, lai izlaistu tabulas indeksu, un `float_format="%.1f"`, lai masas un procentus ierakstītu ar vienu zīmi aiz punkta.

````python
df.to_csv("p6rezultati.csv", index=False, float_format="%.1f")
print("Rezultāti saglabāti datnē p6rezultati.csv.")
````
